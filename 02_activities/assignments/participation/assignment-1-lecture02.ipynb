# Lecture 2 - Jan 20, 2026 #

#### Visualization #1: US Gun Killings in 2018 ####

What can we learn?
This visualization emphasizes the overall human impact of gun violence in a single year, highlighting both the number of deaths and the cumulative “stolen years.”

Is it objective and neutral?
Not fully. The dramatic design, dark background, and flowing arcs are intentionally emotional and guide the viewer toward a strong reaction.

#### Visualization #2: Washington Post Active Shooters ####

What can we learn?
This bar chart clearly shows how the number of active shooter incidents has changed over time, making trends and year-to-year differences easy to see.

Is it objective and neutral?
Yes, more so than the first graph actually. The simple bar chart, consistent scale, and minimal styling focus on clarity and comparison rather than emotional impact, making it closer to an objective, neutral presentation.
