```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75, 50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
# Single plot (one axis)
fig, ax = plt.subplots(figsize=(5, 3))

# Two plots side by side (two axes)
fig, (ax1, ax2) = plt.subplots(
    ncols=2,
    nrows=1,
    figsize=(7, 3)
)

ax1.scatter(x1, y1)
ax2.bar(x2, y2)

fig.show()
```

```python
fig, someaxes = plt.subplot_mosaic(
    [['ax1', 'ax3'],
     ['ax2', 'ax3']],
    figsize=(7, 4)
)

someaxes["ax1"].scatter(x1, y1)
someaxes["ax2"].bar(x2, y2)
someaxes["ax3"].plot(x1, y1)

plt.show()
```

```python
fig, someaxes = plt.subplot_mosaic(
    [['ax1', 'ax3'],
     ['ax2', 'ax3']],
    figsize=(7, 4)
)

someaxes["ax1"].scatter(x1, y1)
someaxes["ax2"].bar(x2, y2)
someaxes["ax3"].plot(x1, y1)

someaxes["ax1"].set_xlabel('A Big Label', fontsize=18)
someaxes["ax2"].set_xlabel('Another Label', fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=18)
```

```python
fig, someaxes = plt.subplot_mosaic(
    [['ax1', 'ax3'],
     ['ax2', 'ax3']],
    figsize=(7, 4),
    layout="constrained"
)

someaxes["ax1"].scatter(x1, y1)
someaxes["ax2"].bar(x2, y2)
someaxes["ax3"].plot(x1, y1)

someaxes["ax1"].set_xlabel('A Big Label', fontsize=18)
someaxes["ax2"].set_xlabel('Another Label', fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=18)
```

```python
# first make our sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(7, 3))

# now call both bar and plot elements on the same axes
ax.bar(x, y1, color="indigo")
ax.plot(x, y2, color="red")
```

```python
# calculate standard deviation of the data
y2_sd = np.std(y2)

# plot the line as before
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color="red")

fig, ax = plt.subplots(figsize=(7, 3))
ax.errorbar(
    x, y2,
    yerr=y2_sd,
    color="red",
    fmt='-o',
    capsize=5
)
```

```python
# calculate standard deviation
y2_sd = np.std(y2)

# plot the error bars only (no data points)
fig, ax = plt.subplots(figsize=(7, 3))
ax.errorbar(
    x,      # x values
    y2,     # y values
    yerr=y2_sd,
    fmt="none"
)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 3))

# plot the main line
ax.plot(x, y2, color="red")

# add customized error bars
ax.errorbar(
    x,
    y2,
    yerr=y2_sd,
    fmt="none",
    ecolor="indigo",
    elinewidth=4,
    capsize=6,
    capthick=4
)

plt.show()
```

```python
from PIL import Image          # open images
import requests               # get images from URLs
from io import BytesIO         # store image bytes
import matplotlib.pyplot as plt

# Get image from the internet
headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(
    "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png",
    headers=headers
)
image_file = BytesIO(response.content)
image = Image.open(image_file)

# Display the image
plt.imshow(image)
plt.axis("off")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Basic line plot
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color="red")

# Add a new axis on top of the existing plot to hold the image
ax_image = fig.add_axes([
    0.10,  # x position (figure coordinates)
    0.11,  # y position (figure coordinates)
    0.15,  # width
    0.35   # height
])

# Show the image inside the new axis
ax_image.imshow(image)
ax_image.axis("off")  # remove ticks/border

plt.show()
```
